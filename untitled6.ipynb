```python
# --- STEP 3: LLM FINE-TUNING AND EVALUATION SETUP ---

# Ensure required libraries are installed and imported for this session
!pip install -q evaluate
!pip install -q rouge-score

import numpy as np
import nltk
import torch # Added import for torch
from datasets import load_dataset # Added load_dataset
from transformers import DataCollatorForSeq2Seq, AutoModelForSeq2SeqLM, Seq2SeqTrainingArguments, Seq2SeqTrainer, AutoTokenizer # Added AutoTokenizer
import evaluate

# Download the necessary NLTK resource for sentence tokenization
nltk.download('punkt_tab')

# Define model checkpoint again to ensure it's in scope
model_checkpoint = "t5-small"

# Initialize the tokenizer (re-added for robustness)
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

# --- Data loading and preprocessing (Integrated from Segment 2 for robustness) ---
# Define model and data parameters (re-defining for local scope)
DATASET_NAME = "cnn_dailymail"
DATASET_CONFIG = "3.0.0"

# Load the dataset from Hugging Face
raw_datasets = load_dataset(DATASET_NAME, DATASET_CONFIG)

# Define T5-specific parameters
prefix = "summarize: "
max_input_length = 512
max_target_length = 128

def preprocess_function(examples):
    inputs = [prefix + doc for doc in examples["article"]]
    model_inputs = tokenizer(inputs, max_length=max_input_length, truncation=True)
    labels = tokenizer(text_target=examples["highlights"], max_length=max_target_length, truncation=True)
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# Apply preprocessing to the entire dataset
tokenized_datasets = raw_datasets.map(preprocess_function, batched=True)

# Select a very small subset for quick, functional training
NUM_TRAIN_EXAMPLES = 500
NUM_EVAL_EXAMPLES = 50

small_train_dataset = tokenized_datasets["train"].shuffle(seed=42).select(range(NUM_TRAIN_EXAMPLES))
small_eval_dataset = tokenized_datasets["validation"].shuffle(seed=42).select(range(NUM_EVAL_EXAMPLES))
# --- End of Data loading and preprocessing ---

# Load the pre-trained T5 model
model = AutoModelForSeq2SeqLM.from_pretrained(model_checkpoint)

# Load the ROUGE metric
metric = evaluate.load("rouge")

def compute_metrics(eval_pred):
    # Fixed, robust implementation for ROUGE calculation
    predictions, labels = eval_pred

    # Decode predictions and labels
    decoded_preds = tokenizer.batch_decode(predictions, skip_special_tokens=True)

    # Replace -100 in labels as these are padding tokens for loss calculation
    labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)

    # ROUGE requires sentence-level newlines for proper comparison
    decoded_preds = ["\n".join(nltk.sent_tokenize(pred.strip())) for pred in decoded_preds]
    decoded_labels = ["\n".join(nltk.sent_tokenize(label.strip())) for label in decoded_labels]

    # Compute ROUGE scores (rouge1, rouge2, rougeL)
    result = metric.compute(predictions=decoded_preds, references=decoded_labels, use_stemmer=True)

    # Round the results to 4 decimal places and convert to percentages
    result = {k: round(v * 100, 4) for k, v in result.items()}

    # Add generated length for monitoring
    prediction_lens = [np.count_nonzero(pred != tokenizer.pad_token_id) for pred in predictions]
    result["gen_len"] = np.mean(prediction_lens)

    return result

# Define training arguments (optimized for Colab T4 GPU)
OUTPUT_DIR = "./final_t5_summarizer_project"

training_args = Seq2SeqTrainingArguments(
    output_dir=OUTPUT_DIR,
    num_train_epochs=3,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    warmup_steps=100,
    weight_decay=0.01,
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    fp16=True if torch.cuda.is_available() else False,
    predict_with_generate=True, # Added to ensure generated tokens are passed to compute_metrics
    report_to="none",
)

# Initialize Data Collator for Seq2Seq
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

# Initialize the Trainer
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=small_train_dataset,
    eval_dataset=small_eval_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator, # Explicitly pass the data collator
    compute_metrics=compute_metrics
)

# Start Fine-Tuning
print("\nSegment 3: Starting Fine-Tuning...")
trainer.train()

# Save the fine-tuned model and tokenizer to a final directory
trainer.save_model(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"\nFine-Tuning Complete. Model saved to: {OUTPUT_DIR}")
```
